# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

De tweede is het meest waarschijnlijk. 

Reden een klopt niet, omdat het niet mogelijk is bij deze relatief lage temperaturen om de moleculen om te zetten naar energie. Deze mengen met andere stoffen in een chemische reactie waarbij energie vrijkomt samen met zuurstof vermengd met andere stoffen. Een vacuüm zal dus nooit ontstaan.

Reden drie klopt niet (volledig), omdat we gekeken hebben naar het ontstaan van van waterdamp en dit lang niet genoeg invloed had om de oorzaak te kunnen zijn. Koolstofdioxide ontstaat wel in het bekerglas, want de vlam dooft uiteindelijk doordat het geen zuurstof meer heeft, maar pas op het moment dat de vlam uit is stijgt het water. 

Daarom klopt reden twee wel, doordat de omgeving om de kaars wordt verhit wordt de druk hoger en ontsnapt de lucht via de onderkant van de maatbeker en komen er bellen aan de rand van het glas. Als de kaars vervolgens uit is koelt de lucht weer af waardoor deze ook minder plek in zal nemen met als resultaat een afname van druk. Door atmosferische druk stijgt daarom uiteindelijk het waterniveau binnen de beker en ontstaat er een evenwicht.
